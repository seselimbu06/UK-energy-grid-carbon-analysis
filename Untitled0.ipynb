```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the CSV. Let pandas automatically detect the header row.
df = pd.read_csv('gb_carbon_intensity.csv')

# Ensure column names are exactly what we expect (in case the file header is named 'index' instead of 'index_rating')
df.columns = ['datetime', 'forecast', 'actual', 'index_rating']

# 2. Remove the header row if it was accidentally read as a data row
df = df[df['datetime'] != 'datetime']

# 3. Convert datetime to actual datetime objects (coerce ignores any weird formatting errors)
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')

# 4. Clean the 'actual' column: force errors to NaN (handles any blank spots like ',,')
df['actual'] = pd.to_numeric(df['actual'], errors='coerce')

# 5. Drop any rows where datetime or actual data is missing/invalid
df = df.dropna(subset=['datetime', 'actual'])

print("✅ Data loaded and cleaned successfully!")
print(f"Total clean rows: {len(df)}")
print("\nFirst 5 rows:")
print(df.head())

```

```python
# 1. Extract hour and day of week (0=Monday, 6=Sunday)
df['hour'] = df['datetime'].dt.hour
df['day_of_week'] = df['datetime'].dt.dayofweek

# 2. Define 'Peak' as Weekdays (Mon-Fri) between 4 PM (16:00) and 9 PM (21:00)
df['is_peak'] = (df['day_of_week'] < 5) & (df['hour'] >= 16) & (df['hour'] <= 21)

# 3. Calculate average actual carbon intensity
peak_avg = df[df['is_peak'] == True]['actual'].mean()
offpeak_avg = df[df['is_peak'] == False]['actual'].mean()
percent_higher = ((peak_avg - offpeak_avg) / offpeak_avg) * 100

# 4. NEW: Analyse the 'index_rating' column!
# Count how many times the grid was rated 'high' or 'very high' in each period
df['is_high_risk'] = df['index_rating'].isin(['high', 'very high'])

peak_high_risk_pct = df[df['is_peak'] == True]['is_high_risk'].mean() * 100
offpeak_high_risk_pct = df[df['is_peak'] == False]['is_high_risk'].mean() * 100

# Print the results
print("=" * 60)
print("📊 UK GRID CARBON INTENSITY ANALYSIS RESULTS")
print("=" * 60)
print(f"Average PEAK carbon intensity:    {peak_avg:.1f} gCO2/kWh")
print(f"Average OFF-PEAK carbon intensity: {offpeak_avg:.1f} gCO2/kWh")
print(f"🔴 Peak hours have {percent_higher:.1f}% HIGHER average carbon intensity")
print("-" * 60)
print(f"🚨 RATING ANALYSIS:")
print(f"During PEAK hours, the grid was rated 'High' or 'Very High' {peak_high_risk_pct:.1f}% of the time.")
print(f"During OFF-PEAK hours, it was rated 'High' or 'Very High' only {offpeak_high_risk_pct:.1f}% of the time.")
print("=" * 60)
```

```python
# Set a clean, professional plotting style
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(8, 5))

# Data for the bar chart
categories = ['Off-Peak Hours', 'Peak Hours']
values = [offpeak_avg, peak_avg]
colors = ['#4CAF50', '#F44336'] # Green for off-peak, Red for peak

# Create the bars
bars = plt.bar(categories, values, color=colors, width=0.6)

# Add title and labels
plt.title('UK Grid Carbon Intensity: Peak vs. Off-Peak Hours', fontsize=14, fontweight='bold')
plt.ylabel('Average Carbon Intensity (gCO2/kWh)', fontsize=12)
plt.ylim(0, max(values) * 1.2)

# Add the exact numbers on top of each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 5, f'{yval:.1f}',
             ha='center', va='bottom', fontsize=12, fontweight='bold')

# Save and show
plt.tight_layout()
plt.savefig('carbon_intensity_chart.png', dpi=300)
print("✅ Chart saved as 'carbon_intensity_chart.png'")
plt.show()
```
